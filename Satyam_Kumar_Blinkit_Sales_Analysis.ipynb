This notebook performs an end-to-end exploratory data analysis of an educational quick-commerce sales dataset.

## 1. Objective

- Measure orders, units, revenue and profit.
- Compare category and city performance.
- Analyze monthly revenue trends.
- Examine customer ratings and delivery performance.
- Produce business-oriented observations.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

pd.set_option('display.max_columns', None)
DATA_FILE = 'blinkit_sales_dataset_educational.csv'
df = pd.read_csv(DATA_FILE, parse_dates=['Order_Date'])
df.head()
```

```python
# Basic structure and data-quality checks
print('Shape:', df.shape)
print('\nMissing values:\n', df.isna().sum())
print('\nDuplicate rows:', df.duplicated().sum())
df.describe(include='all').T
```

## 2. Key Performance Indicators

```python
kpis = pd.Series({
    'Total Orders': df['Order_ID'].nunique(),
    'Units Sold': int(df['Quantity'].sum()),
    'Revenue': round(df['Sales'].sum(), 2),
    'Profit': round(df['Profit'].sum(), 2),
    'Average Order Value': round(df['Sales'].mean(), 2),
    'Average Rating': round(df['Customer_Rating'].mean(), 2),
    'Average Delivery Minutes': round(df['Delivery_Minutes'].mean(), 2)
})
kpis
```

## 3. Category Analysis

```python
category_summary = df.groupby('Category').agg(
    Orders=('Order_ID','nunique'),
    Units=('Quantity','sum'),
    Revenue=('Sales','sum'),
    Profit=('Profit','sum'),
    Avg_Rating=('Customer_Rating','mean')
).sort_values('Revenue', ascending=False)
display(category_summary)

category_summary['Revenue'].sort_values().plot(kind='barh', figsize=(9,5))
plt.title('Revenue by Category')
plt.xlabel('Revenue')
plt.tight_layout()
plt.show()
```

## 4. City Analysis

```python
city_summary = df.groupby('City').agg(
    Orders=('Order_ID','nunique'),
    Revenue=('Sales','sum'),
    Profit=('Profit','sum'),
    Avg_Delivery=('Delivery_Minutes','mean')
).sort_values('Revenue', ascending=False)
display(city_summary)

city_summary['Revenue'].sort_values().plot(kind='barh', figsize=(9,5))
plt.title('Revenue by City')
plt.xlabel('Revenue')
plt.tight_layout()
plt.show()
```

## 5. Monthly Revenue Trend

```python
monthly = (df.assign(Month=df['Order_Date'].dt.to_period('M').astype(str))
             .groupby('Month')
             .agg(Revenue=('Sales','sum'), Profit=('Profit','sum'), Orders=('Order_ID','nunique'))
             .reset_index())
display(monthly)

plt.figure(figsize=(9,5))
plt.plot(monthly['Month'], monthly['Revenue'], marker='o')
plt.title('Monthly Revenue Trend')
plt.xlabel('Month')
plt.ylabel('Revenue')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 6. Customer & Delivery Analysis

```python
print('Payment method distribution:')
display(df['Payment_Method'].value_counts().to_frame('Orders'))

print('Customer type distribution:')
display(df['Customer_Type'].value_counts().to_frame('Orders'))

print('Average rating:', round(df['Customer_Rating'].mean(), 2))
print('Average delivery time:', round(df['Delivery_Minutes'].mean(), 2), 'minutes')
```

## 7. Business Insights

1. High-revenue categories can be prioritized for inventory availability and assortment planning.
2. City-level revenue should be considered together with delivery time before operational expansion.
3. Monthly trends can help identify demand changes and periods requiring further investigation.
4. Ratings and delivery metrics can be monitored together as customer-experience indicators.
5. New versus returning customer behavior can be used in future retention analysis.
